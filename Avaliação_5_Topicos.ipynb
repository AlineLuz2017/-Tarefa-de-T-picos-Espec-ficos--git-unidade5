```python
# Execução completa do sistema gratuito
import os
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings

# 1. Criação do arquivo de texto base sobre Saúde Mental e Tecnologia
conteudo_tecnologia = """
O uso excessivo de tecnologias e redes sociais tem gerado impactos significativos na saúde mental moderna,
incluindo o fenômeno do FOMO (medo de estar perdendo algo) e a ansiedade gerada pela comparação social constante.
A exposição prolongada a telas, especialmente antes de dormir, interfere diretamente na produção de melatonina,
prejudicando a qualidade do sono e aumentando os níveis de fadiga mental e estresse crônico.
O "detox digital" ou pausas programadas longe de smartphones e computadores são práticas essenciais para recuperar o foco,
reduzir a exaustão cognitiva e promover o bem-estar emocional.
Estabelecer limites claros para o uso de notificações, reservar momentos livres de telas durante as refeições e
priorizar interações presenciais são estratégias fundamentais para manter um equilíbrio saudável após a era digital.
Práticas de atenção plena (mindfulness) e atividades ao ar livre ajudam a resgatar a presença no momento presente e
a diminuir a dependência psicológica dos dispositivos eletrônicos.
"""

with open("base_saude_tecnologia.txt", "w", encoding="utf-8") as f:
    f.write(conteudo_tecnologia)

# 2. Carregamento e divisão do documento em chunks
loader = TextLoader("base_saude_tecnologia.txt", encoding="utf-8")
documentos = loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)
chunks = text_splitter.split_documents(documentos)
print(f"Total de chunks gerados: {len(chunks)}")

# 3. Geração de Embeddings e Banco Vetorial FAISS (100% local e gratuito)
print("Gerando embeddings e criando banco FAISS...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vectorstore = FAISS.from_documents(chunks, embeddings)
print("Banco FAISS criado com sucesso!\n")

# 4. Sistema de Recuperação de Informação (RAG)
pergunta = "O que é detox digital e qual a sua importância?"

# Busca o trecho mais relevante usando similaridade de vetores
docs_relacionados = vectorstore.similarity_search(pergunta, k=1)

# 5. Exibindo o resultado
print("="*60)
print(f"Pergunta: {pergunta}")
print("="*60)
print("Resposta recuperada do texto base:")
print(docs_relacionados[0].page_content)
print("="*60)
```
